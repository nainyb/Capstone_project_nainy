4. Fuel Type vs. Manufacturer (Heatmap)

To show which manufacturers are offering which types of vehicles.

```python
# Group by Manufacturer and Fuel, then count models
grouped_data = filtered_df.groupby(['Manufacturer', 'Fuel'])['Model'].count()

# Reshape the data for the heatmap using unstack()
heatmap_data = grouped_data.unstack('Fuel').fillna(0) 
```

```python
plt.figure(figsize=(10, 6))  # Adjust figure size for better readability
sns.heatmap(heatmap_data, annot=True, fmt='g', cmap='viridis')  # Customize colormap and annotations
plt.title('Number of Models by Manufacturer and Fuel Type')  # Add a title
plt.xlabel('Fuel Type')  # Add x-axis label
plt.ylabel('Manufacturer')  # Add y-axis label
plt.show()
```

```python
pivot = filtered_df.pivot_table(
    index="Manufacturer", 
    columns="Fuel", 
    values="Model", 
    aggfunc="count", 
    fill_value=0
)

plt.figure(figsize=(12,8))
sns.heatmap(pivot, cmap="Blues", annot=True, fmt="d")
plt.title("Fuel Type Distribution Across Manufacturers")
plt.show()
```
